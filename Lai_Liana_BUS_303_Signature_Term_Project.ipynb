## Deliverable 1: Cleaning and Preparing the Starbucks Menu for Analysis

### Task 1.1: Import Pandas and Load the Starbucks CSV

```python
import pandas as pd

sbux = pd.read_csv(
    'BUS303_Signature_Term_Project_Starbucks_Beverage_Menu.csv')

print('Loaded shape:', sbux.shape)
```

### Task 1.2: Explore Rows with Blank Values

```python
sbux.isnull().sum()
```

### Task 1.3: View the Dataframe with the Head Function

```python
sbux.head()
```

### Task 1.4: Remove Rows with All Blank Entries

```python
sbux2 = sbux.dropna(how='all').copy()

print(
    'Shape after removing fully blank rows:',
    sbux2.shape
)
```

### Task 1.5: Validate That the Blank Rows Were Removed

```python
print(
    'Fully blank rows remaining:',
    sbux2.isnull().all(axis=1).sum()
)

sbux2.isnull().sum()
```

### Task 1.6: Explore the Data Types

```python
sbux2.dtypes
```

### Task 1.7: Identify the Incorrect Data Type

```python
print(
    'Sugar_g sample values:',
    sbux2.iloc[:, 13].dropna().unique()[:10]
)

print(
    'Sugar_g data type:',
    sbux2.iloc[:, 13].dtype
)
```

### Task 1.8: Change Sugar_g to a Float

```python
sbux2['Sugar_g'] = sbux2['Sugar_g'].replace({
    'zero': '0'
})

sbux1 = sbux2.copy()

sbux1['Sugar_g'] = sbux1['Sugar_g'].astype(float)

print(
    'Sugar_g data type after conversion:',
    sbux1['Sugar_g'].dtype
)
```

### Task 1.9: Validate the Sugar_g Conversion

```python
print(
    'Sugar_g data type:',
    sbux1['Sugar_g'].dtype
)

print(
    'Values that cannot be converted:',
    pd.to_numeric(
        sbux1['Sugar_g'],
        errors='coerce'
    ).isna().sum()
)
```

### Task 1.10: Explore the Remaining NaN Values

```python
sbux1.info()
```

### Task 1.11: Remove the Row Containing the NaN Value

```python
df2 = sbux1.dropna().copy()

sbux3 = df2.copy()

print(
    'Shape after removing rows with missing values:',
    sbux3.shape
)
```

### Task 1.12: Validate That the NaN Value Was Removed

```python
sbux3.info()

print(
    'Missing values remaining:',
    sbux3.isnull().sum().sum()
)
```

### Task 1.13: Correct the Caffeine Column Name

sbux3 = sbux3.rename(
    columns={
        'Caffiene_milligrams': 'Caffeine_milligrams'
    }
)

### Task 1.14: Validate the Corrected Column Name

```python
sbux3.info()

print(
    'Caffeine_milligrams present:',
    'Caffeine_milligrams' in sbux3.columns
)

print(
    'Incorrect spelling remains:',
    'Caffiene_milligrams' in sbux3.columns
)
```

### Task 1.15: Export the Cleaned Dataframe to sbux4.csv

```python
sbux3.to_csv(
    'sbux4.csv',
    index=False
)

print('Saved sbux4.csv')
```

### Task 1.16: Reload and Validate sbux4.csv

```python
sbux4 = pd.read_csv('sbux4.csv')

print(
    'Reloaded shape:',
    sbux4.shape
)

print(
    'Column names match:',
    sbux4.columns.tolist() == sbux3.columns.tolist()
)

print(
    'Data values match:',
    sbux4.equals(sbux3.reset_index(drop=True))
)

print(
    'Missing values in reloaded file:',
    sbux4.isnull().sum().sum()
)

sbux4.head()
```
